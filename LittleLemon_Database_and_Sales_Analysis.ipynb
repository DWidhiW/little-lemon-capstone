```python
import sqlite3

try:
    # 1. Connect to SQLite database (creates a local file in Colab)
    connection = sqlite3.connect("little_lemon.db")
    cursor = connection.cursor()
    print("Successfully connected to the Little Lemon SQLite database\n")

    # Create tables for demonstration
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS Orders (
            OrderID INT PRIMARY KEY,
            Quantity INT
        )
    """)
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS Bookings (
            BookingID INT PRIMARY KEY,
            BookingDate DATE,
            TableNumber INT
        )
    """)
    connection.commit()

    # 2. Example: GetMaxQuantity logic
    print("--- GetMaxQuantity Test ---")
    cursor.execute("SELECT MAX(Quantity) FROM Orders")
    print(f"Max Quantity: {cursor.fetchall()}\n")

    # 3. Example: AddBooking logic
    print("--- AddBooking Test ---")
    cursor.execute("INSERT OR REPLACE INTO Bookings (BookingID, BookingDate, TableNumber) VALUES (99, '2026-10-20', 5)")
    connection.commit()
    print("Booking added successfully.\n")

except Exception as e:
    print(f"Error: {e}")

finally:
    if 'connection' in locals():
        connection.close()
        print("Database connection is closed.")
```
